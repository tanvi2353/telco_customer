```python
# columns where 0 is not valid medical value
# glucose,bloodpressure,skinthickness,insulin,bmi
```

```python
# columns where 0 is not valid medical values
# glucose,bloodpressure,skinthickness,insulin,bmi

invalid_cols = ['glucose',
                'bloodpressure',
                'skinthickness',
                'insulin',
                'bmi'
]

# replace 0 with nan
df[invalid_cols] = df[invalid_cols].replace(0,np.nan)
```

```python
# impute missing values
for col in invalid_cols:
  df[col].fillna(df[col].median(),inplace=True)

#medium represents a typical patient not too high not to low
# not influenced by outliers
```

```python
df.isnull().sum()
```

```python
# feature and target separation
# feature
x=df.drop(columns=['outcome'])
# target
y=df['outcome']

```

```python
x.columns
```

```python
y.name
```

```python
# train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,stratify=y,random_state=42)
```

```python
y.value_counts(normalize=True)
```

```python
# feature scaling preprocessing
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train[:10]
```

```python
# here we will use logestic regression bcz it is a,binary
# classification problem
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
# prediction
y_pred_train=model.predict(x_train)
y_pred_test=model.predict(x_test)
```

```python
y_pred_train[:5]
```

```python
y_pred_test[:5]
```

```python
y_test[:5]
```

```python
# score
model.score(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
# confusion matrix
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred_test)
```

```python
# classification report
from sklearn.metrics import classification_report
cr=(classification_report(y_test,y_pred_test))
print(cr)
```

```python
# next model we using here is SVM(support vector machine)
# it finds a strong,margin between two classes
# it finds a strong margin based decision boundary that often
# improves recall on small,well scaled dataset
```

```python
from sklearn.svm import SVC
svc_model = SVC()
svc_model.fit(x_train,y_train)
```
